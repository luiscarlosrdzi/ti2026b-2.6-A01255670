**Pregunta:** ¿Existe relación entre las horas de estudio y la calificación en un conjunto de datos sintético?

**Procedencia de datos:** Hice uso de datos sintéticos los cuales fueron generados dentro del notebook con una semilla fija.

**Salida esperada:** Una tabla resumen, el coeficiente de correlación y una gráfica de dispersión.

```python

import sys
print("Python:", sys.version)

# Librerías usadas en el análisis
import numpy as np               
import pandas as pd              
import matplotlib.pyplot as plt 

# Versiones usadas
print("numpy:", np.__version__)
print("pandas:", pd.__version__)
```

```python
# Generación de datos sintéticos
# Crea 100 estudiantes ficticios: las calificaciones dependen de las horas
# de estudio más un poco de ruido aleatorio.

rng = np.random.default_rng(42)  # semilla fija: siempre salen los mismos datos
n = 100                          # número de estudiantes simulados

# Horas de estudio entre 0 y 10
horas = rng.uniform(0, 10, n)

# Calificación = base 50 + 4.5 puntos por hora + ruido; se limita al rango 0-100
calificacion = np.clip(50 + 4.5 * horas + rng.normal(0, 6, n), 0, 100)

# Se guardan en una tabla para analizarlos
df = pd.DataFrame({"horas_estudio": horas, "calificacion": calificacion})
df.head() 
```

```python
# Análisis de los datos
print(df.describe().round(2))

# Coeficiente de correlación
r = df["horas_estudio"].corr(df["calificacion"])
print(f"\nCorrelación de Pearson: {r:.3f}")

# Gráfica de dispersión: cada punto es un estudiante
plt.scatter(df["horas_estudio"], df["calificacion"], alpha=0.6)
plt.xlabel("Horas de estudio (sintético)")
plt.ylabel("Calificación (sintético)")
plt.title("Horas de estudio vs. calificación")
plt.show()
```

## Conclusión

La correlación entre horas de estudio y calificación fue de 0.906, lo cual nos dice que tiene una relacion muy fuerte, por lo cual es muy grande el impaco de las horas de estudio con la calificacion, dandonos como resultado entre mas las horas de estudio pues obtedran una mayor calificacion.
